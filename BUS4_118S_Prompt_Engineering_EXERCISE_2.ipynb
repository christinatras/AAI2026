```python
# Exercise 2
# ReACT Code Generation

import pandas as pd

required_columns = ["math_score", "physics_score", "english_score"]

try:
# Read the CSV file
    df = pd.read_csv("student_scores.csv", skiprows=1)

# Check if dataset is empty
    if df.empty:
        print("The dataset is empty. No analysis can be performed.")

    else:
# Check if required columns exist
        missing_columns = [col for col in required_columns if col not in df.columns]

        if missing_columns:
            print("Missing required columns:", missing_columns)

        else:
            print("Cleaned Student Scores Dataset:")
            print(df)
            print()

            for subject in required_columns:
# Remove missing values before calculations
                clean_scores = pd.to_numeric(df[subject], errors="coerce").dropna()

                if clean_scores.empty:
                    print(f"{subject} has no valid data.")
                    print()
                else:
                    print(f"{subject} statistics:")
                    print("Average:", clean_scores.mean())
                    print("Maximum:", clean_scores.max())
                    print("Minimum:", clean_scores.min())
                    print("Count above 90:", (clean_scores > 90).sum())
                    print()

except FileNotFoundError:
    print("Error: student_scores.csv was not found.")

except pd.errors.EmptyDataError:
    print("Error: The CSV file is empty.")

except Exception as e:
    print("An unexpected error occurred:", e)
```
