```python
# EXERCISE 1
# Target Support Customer Service Chatbot

# Step 1: Classify Issue of Chatbot and Purpose, Target Support
def classify_issue():
    print("\nTarget Support: How can I help you today?")
    print("1. Order Tracking")
    print("2. Account Help")
    print("3. Billing Issue")
    print("4. Refund Request")
    print("5. Technical Support")

    choice = input("Enter the number of your issue: ").strip()

    if choice == "1":
        return "order_tracking"
    elif choice == "2":
        return "account_help"
    elif choice == "3":
        return "billing_issue"
    elif choice == "4":
        return "refund_request"
    elif choice == "5":
        return "technical_support"
    else:
        print("Target Support: Sorry, I didn't understand that.")
        return "menu"

# Step 2: Gather information depending on the issue type
# Order number and email validation to ensure numbers and emails contain "@"

def gather_missing_info(issue):
    info = {}

    def get_order_number():
        while True:
            order = input("Target Support: Please enter your order number: ").strip()
            if order.isdigit():
                return int(order)
            else:
                print("Target Support: Invalid order number. Please enter numbers only.")

    def get_email():
        while True:
            email = input("Target Support: Please enter your email: ").strip()
            if "@" in email:
                return email
            else:
                print("Target Support: Invalid email. Please include an '@' symbol.")


    if issue == "order_tracking":
        info["order_number"] = get_order_number()
        info["email"] = get_email()

    elif issue == "account_help":
        info["email"] = get_email()

    elif issue == "billing_issue":
        info["email"] = get_email()
        info["problem"] = input("Target Support: Describe the billing issue: ")

    elif issue == "refund_request":
        info["order_number"] = get_order_number()
        info["email"] = get_email()
        info["reason"] = input("Target Support: Why are you requesting a refund?: ")

    elif issue == "technical_support":
        info["email"] = get_email()
        info["device"] = input("Target Support: What device are you using?: ")
        info["error"] = input("Target Support: Describe the technical issue: ")

    return info

# Based on selected issue, Chatbot gives automated response

def propose_solution(issue, info):
    if issue == "order_tracking":
        print(f"Target Support: Your order {info['order_number']} is currently in transit and should arrive within 3-5 days.")

    elif issue == "account_help":
        print("Target Support: Please try resetting your password using the password reset link.")

    elif issue == "billing_issue":
        print("Target Support: Please check if the charge is pending. If it remains after 24 hours we can escalate.")

    elif issue == "refund_request":
        print("Target Support: Your refund request has been recorded and will be reviewed.")

    elif issue == "technical_support":
        print("Target Support: Please restart the application or clear your browser cache.")

# Step 4: Ask user if they want to escalate to a live agent

def live_agent_prompt():
    response = input("Target Support: Would you like to speak to a live agent? (yes/no): ").lower().strip()

    if response == "yes":
        print("Target Support: Redirecting to live support agent..")
        return True
    elif response == "no":
        print("Target Support: Returning to the main menu.")
        return False
    else:
        print("Target Support: Please answer yes or no.")
        return live_agent_prompt()

# Live agent response is specific to each issue

def live_agent_chat(issue, info):
    print("\nLive Agent: Hi, my name is Sarah. I’ll be assisting you today.")
    customer_issue = input("Customer: ")

    email = info.get("email", "your email")

    if issue == "order_tracking":
        print("Live Agent: Thank you for sharing that. I have documented your order tracking concern.")
        print(f"Live Agent: We will send an update about your order status to {email}.")

    elif issue == "account_help":
        print("Live Agent: Thank you for sharing that. I have documented your account support issue.")
        print(f"Live Agent: Our team will follow up with next steps at {email}.")

    elif issue == "billing_issue":
        print("Live Agent: Thank you for sharing that. I have documented your billing concern for review.")
        print(f"Live Agent: A billing specialist will follow up with you at {email}.")

    elif issue == "refund_request":
        print("Live Agent: Thank you for sharing that. I have documented your refund request.")
        print(f"Live Agent: We will send the refund review details to {email}.")

    elif issue == "technical_support":
        print("Live Agent: Thank you for sharing that. I have documented your technical support issue.")
        print(f"Live Agent: Our technical team will send troubleshooting steps to {email}.")

    print("Live Agent: Please expect a response from Target by email within 1-2 hours.")

# If more concerns, it will be noted in the email that is sent
# Prompts user to continue conversation via email

    follow_up = input("Live Agent: Is there anything else I can help you with?\nCustomer: ").strip().lower()

    if follow_up == "nothing else":
        print("Live Agent: Thank you for contacting Target Support.")
        print("Target Support: This chat has ended.")

    elif follow_up == "no":
        print("Target Support: Thank you for contacting Target Support, this chat has ended.")

    else:
        print("Live Agent: Thank you for sharing your additional concern.")
        print(f"Live Agent: I have added it to your case and our team will follow up at {email}.")
        print("Live Agent: Please expect a response from Target by email within 1-2 hours.")
        print("Target Support: Thank you for contacting Target Support, this chat has ended.")

# Main function controls overall Chatbot conversation flow
# Classification of Chatbot - info gathering, solutions, escalation

def main():
    while True:
        issue = classify_issue()

        if issue == "menu":
            continue

        info = gather_missing_info(issue)
        propose_solution(issue, info)

        escalate = live_agent_prompt()

        if escalate:
            live_agent_chat(issue, info)
            break


if __name__ == "__main__":
    main()
```
